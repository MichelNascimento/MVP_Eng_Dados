```python
import re

# Caminhos dos arquivos
caminho_alarmes = "/Volumes/workspace/default/dados_brutos_mvp/alarmes.parquet"
caminho_manobras = "/Volumes/workspace/default/dados_brutos_mvp/registro_manobras.parquet"

# Lendo os arquivos
df_alarmes = spark.read.parquet(caminho_alarmes)
df_manobras = spark.read.parquet(caminho_manobras)

# Função para limpar os nomes das colunas (troca espaços e parênteses por _)
def limpar_nomes_colunas(df):
    # O regex abaixo procura os caracteres proibidos e substitui por "_"
    novas_colunas = [re.sub(r'[ ,;{}()\n\t=]', '_', c) for c in df.columns]
    return df.toDF(*novas_colunas)

df_alarmes = limpar_nomes_colunas(df_alarmes)
df_manobras = limpar_nomes_colunas(df_manobras)

# Salvando os dados como tabelas na Camada Bronze
df_alarmes.write.format("delta").mode("overwrite").saveAsTable("workspace.default.bronze_alarmes")
df_manobras.write.format("delta").mode("overwrite").saveAsTable("workspace.default.bronze_manobras")

print("Sucesso! Tabelas da Camada Bronze criadas e salvas no banco de dados.")
```
